```python
from sklearn.model_selection import train_test_split
X, y = olive.data, olive.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=150)
X1D = pca.fit_transform(X_train)
x_trans = pca.transform(X_test)
```

Now the next thing is clustering image using k-means

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
kmeans = KMeans(n_clusters = 10, n_init = 10, random_state=42)
kmeans.fit(X1D)
```

```python
dingo = silhouette_score(X1D, kmeans.labels_)
```

```python
import matplotlib.pyplot as plt
import numpy as np 
k_range = range(5, 150, 5)
best_index = np.argmax(dingo)
best_k = k_range[best_index]
best_score = dingo[best_index]



plt.figure(figsize=(9,11))

plt.xlabel("$k$")
plt.ylabel("Silhouette Score")
plt.plot(k_range, dingo, 'bo-')
plt.grid()
plt.show()

```
